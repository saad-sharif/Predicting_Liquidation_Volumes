```python
%%time
# Merge multiple coins into one master dataset

# Configuration
coins = ["ltc","btc","eth","sol","xrp","doge","ada","link","avax","bch",]
all_master_dfs = []

# Loop over the coins
for coin in coins:

    print("\n================================================================================")
    print(f"Processing coin: {coin.upper()}")
    print("=================================================================================")

    files = {"microstructure":f"{coin}_microstructure.parquet",
             "orderbook":f"{coin}_orderbook_features_1m.parquet",
             "oi":f"{coin}usdt_oi_features.parquet",
             "funding":f"{coin}_funding.parquet",
             "longshort":f"{coin}_longshort.parquet",
             "basis":f"{coin}_basis.parquet",
             "liquidations":f"{coin}_liquidations.parquet",
             "onchain":f"{coin}_onchain_5m.parquet",
             "social":f"{coin}_social_5m.parquet",}

    # Load data
    dfs = {}

    for name, path in files.items():
        print(f"Loading {path}")
        df = pd.read_parquet(path)

        # timestamp standardization
        timestamp_col = None
        for col in ["timestamp","exchangeTimestamp"]:
            if col in df.columns:timestamp_col = col
                break

        df = df.rename(columns={timestamp_col:"timestamp"})
        df["timestamp"] = pd.to_datetime(df["timestamp"],utc=True)
        if name == "orderbook":df["timestamp"] = (df["timestamp"].dt.floor("1min")) # Align high-resolution timestamps to 1-minute buckets
        df = (df.sort_values("timestamp").drop_duplicates(subset="timestamp",keep="last")) # remove duplicate timestamps
        redundant_cols = [c for c in df.columnsif c in ["instrument","exchange"]] # drop redundant instrument cols
        df = df.drop(columns=redundant_cols,errors="ignore")
        dfs[name] = df

    # Create Master Table
    start = min(df["timestamp"].min()for df in dfs.values())
    end = max(df["timestamp"].max()for df in dfs.values())
    master_index = pd.date_range(start=start,end=end,freq="1min",tz="UTC")
    master_df = pd.DataFrame({"timestamp":master_index})
    print(f"Master timeline for "f"{coin.upper()}: "f"{len(master_df):,} rows")

    # Merge everything
    for name, df in dfs.items():
        print(f"Merging {name}")
        master_df = master_df.merge(df,on="timestamp",how="left")

    # Add coin column
    master_df["coin"] = coin

    # Forward Fill columns
    ffill_cols = ["exchange_balance",
                  "active_addresses_24h",
                  "transaction_volume",
                  "social_volume",
                  "social_dominance",
                  "sentiment_polarity",
                  "abnormal_attention",
                  "whale_tx_count_100k",
                  "whale_tx_count_1m",
                  "open_interest",
                  "funding_rate",
                  "cumulative_funding",
                  "basis",
                  "mark_index_dislocation",
                  "long_short_skew",]

    existing_ffill = [c for c in ffill_colsif c in master_df.columns]
    master_df[existing_ffill] = (master_df[existing_ffill].ffill())

    # Zero fill these features
    zero_fill_cols = ["liquidation_volume","liquidation_acceleration","liquidation_acceleration_5m","liquidation_acceleration_15m","signed_volume",]
    existing_zero = [c for c in zero_fill_colsif c in master_df.columns]
    master_df[existing_zero] = (master_df[existing_zero].fillna(0))
    print(f"\nFinal shape for "f"{coin.upper()}:")
    print(master_df.shape)
    all_master_dfs.append(master_df)

# Append together
combined_master_df = pd.concat(all_master_dfs,ignore_index=True)
combined_master_df = (combined_master_df.sort_values(["coin","timestamp"]).reset_index(drop=True))

# Save combined dataset
combined_master_df.to_parquet("crypto_master_dataset.parquet",index=False)
combined_master_df.to_csv("crypto_master_dataset.csv",index=False)

print("\n================================================================================")
print("FINAL COMBINED SHAPE:")
print(combined_master_df.shape)
print("\nCOINS INCLUDED:")
print(combined_master_df["coin"].value_counts())
print("\nCOLUMNS:")
print(combined_master_df.columns.tolist())
print("\nHEAD:")
print(combined_master_df.head())
print("\nSaved:")
print("crypto_master_dataset.parquet")
print("crypto_master_dataset.csv")
```

```python
%%time
import numpy as np

def grouped_rolling_zscore(df,group_col,value_col,window,min_periods=None):
    if min_periods is None: min_periods = window

    rolling_mean = (df.groupby(group_col)[value_col].transform(lambda x: x.rolling(window=window,min_periods=min_periods).mean()))
    rolling_std = (df.groupby(group_col)[value_col].transform(lambda x: x.rolling(window=window,min_periods=min_periods).std()).replace(0, np.nan))

    return (df[value_col] - rolling_mean) / rolling_std


# CREATE BACKUP COPY
combined_master_df_bkp = combined_master_df.copy()

# PARAMETERS
window_local = 10080 #240

# SORT DATA
combined_master_df_bkp = (combined_master_df_bkp.sort_values(["coin", "timestamp"]).reset_index(drop=True))

# REQUIRED COLUMN CHECK
required_cols = ["coin","timestamp","liquidation_volume",]
missing_cols = [colfor col in required_cols if col not in combined_master_df_bkp.columns]
if missing_cols:raise KeyError(f"Missing columns: {missing_cols}")

# LIQUIDATION VOLUME Z-SCORE
combined_master_df_bkp["liq_vol_zscore"] = grouped_rolling_zscore(df=combined_master_df_bkp,group_col="coin",value_col="liquidation_volume",window=window_local)

# LIQUIDATION Z-SCORE BUCKETS
combined_master_df_bkp["liq_vol_bucket"] = np.select(
    [combined_master_df_bkp["liq_vol_zscore"] >= 10,combined_master_df_bkp["liq_vol_zscore"] >= 5,combined_master_df_bkp["liq_vol_zscore"] >= 2,],
    ["high_liq_vol","medium_liq_vol","low_liq_vol",],default="no_liq_vol")

# OPTIONAL BINARY TARGET
combined_master_df_bkp["liq_vol_flag"] = (combined_master_df_bkp["liq_vol_zscore"] >= 2).astype(int)

# VIEW RELEVANT COLUMNS
liq_vol_zscore_df = combined_master_df_bkp[["coin","timestamp","liquidation_volume","liq_vol_zscore","liq_vol_bucket","liq_vol_flag",]].copy()
liq_vol_zscore_df.head()
```
